```ruby
Trevl::DataSource.register("elections", Trevl::DataSource::Static.new(
  data: {
    "parties" => {
      "data" => [
        {"name" => "Grüne",  "share" => 6},
        {"name" => "AFD",  "share" => 16},
        {"name" => "SPD",  "share" => 11},
        {"name" => "CDU",    "share" => 22},
        {"name" => "Linke",  "share" => 5},
      ]
    }
  }
))

nb = Trevl::Notebook.new

chart = nb.chart(<<~YAML)
  components:
  - id: election_results
    type: chart
    api: elections
    highchartsData:
      chart:
        type: column
      title:
        text: Wahlergebnis - Kommunalwahl in Klein Datenhausen
      series:
      - name: Anteil in %
        data:
          x: "$parties.data.name"
          y: "$parties.data.share"
YAML
```
